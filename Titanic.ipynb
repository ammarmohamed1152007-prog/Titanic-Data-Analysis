```python
df['Survived'].value_counts().plot(kind='bar')
plt.xlabel('Survived')
plt.ylabel('Number of survived')
plt.title('Survived vs Non-Survived')
plt.show()
```

```python
df.plot(x='Age',y='Fare',kind='scatter')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')
plt.show()
```

### Insight: Age and Fare

- The relationship between Age and Fare is not strongly linear.
- Some passengers paid significantly higher fares than others.
- Higher fares are generally associated with higher passenger classes.

```python
df['Age'].plot(kind='hist', bins=20)
plt.xlabel('Age')
plt.title('Age Distribution')
plt.show()
```

### Insight: Age Distribution

- Most passengers were adults.
- The dataset also contains a smaller number of children and older passengers.
- The Age distribution shows that passengers came from different age groups.

```python
df.groupby("Sex")["Survived"].mean().plot(kind="bar")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

### Insight: Survival by Gender

- Female passengers had a higher survival rate than male passengers.
- Gender appears to be an important factor affecting survival.

```python
df.groupby("Pclass")["Survived"].mean().plot(kind="bar")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

### Insight: Survival by Passenger Class

- First-class passengers had a higher survival rate than passengers in lower classes.
- Passenger class appears to be an important factor related to survival.

```python
fig = px.histogram(
    df,
    x="Age",
    color="Sex",
    animation_frame="Pclass",
    title="Passengers by Age and Gender"
)

fig.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

# Key Insights and Conclusions

### Main Findings

1. Female passengers had a higher survival rate than male passengers.
2. Passengers in higher classes generally had better survival chances.
3. Age and Fare do not show a strong direct linear relationship.
4. Most passengers were adults, with fewer children and elderly passengers.
5. Passenger class, gender, and fare appear to be important factors related to survival.

### Conclusion

The Titanic dataset shows that survival was associated with several passenger characteristics, especially gender and passenger class. Data cleaning and exploratory data analysis helped identify patterns and relationships within the dataset.
