```python
df['Survived'].value_counts().plot(
    kind='bar',
    color=['skyblue', 'salmon']
)

plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.title('Survived vs Non-Survived')

plt.show()
```

```python
df.plot(
    x='Age',
    y='Fare',
    kind='scatter',
    color='purple'
)

plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')

plt.show()
```

### Insight: Age and Fare

- The relationship between Age and Fare is not strongly linear.
- Some passengers paid significantly higher fares than others.
- Higher fares are generally associated with higher passenger classes.

```python
df['Age'].plot(
    kind='hist',
    bins=20,
    color='teal'
)

plt.xlabel('Age')
plt.title('Age Distribution')

plt.show()
```

### Insight: Age Distribution

- Most passengers were adults.
- The dataset also contains a smaller number of children and older passengers.
- The Age distribution shows that passengers came from different age groups.

```python
df.groupby("Sex")["Survived"].mean().plot(
    kind="bar",
    color=["skyblue", "salmon"]
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

### Insight: Survival by Gender

- Female passengers had a higher survival rate than male passengers.
- Gender appears to be an important factor affecting survival.

```python
df.groupby("Pclass")["Survived"].mean().plot(
    kind="bar",
    color=["gold", "skyblue", "salmon"]
)

plt.title("Survival Rate by Passenger Class")

plt.xlabel("Passenger Class")

plt.ylabel("Survival Rate")

plt.show()
```

### Insight: Survival by Passenger Class

- First-class passengers had a higher survival rate than passengers in lower classes.
- Passenger class appears to be an important factor related to survival.

```python
fig = px.histogram(
    df,
    x="Age",
    color="Sex",
    animation_frame="Pclass",
    title="Passengers by Age and Gender"
)

fig.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

## 🔗 Correlation Matrix Insight

Survival had a moderate negative correlation with passenger class (-0.34) and a positive correlation with fare (0.26). Passenger class and fare also showed a strong negative correlation (-0.55), indicating that higher-class passengers generally paid higher fares.

```python
plt.plot(df["Age"], color="teal")

plt.fill_between(df.index, df["Age"], color="skyblue", alpha=0.3)

plt.title("Age Distribution")

plt.xlabel("Passenger")

plt.ylabel("Age")

plt.show()
```

## 👤 Age Distribution Insight

Most passengers were young adults, while very old passengers were less common.

```python
sns.barplot(x="Sex", y="Survived", data=df, palette=["skyblue", "salmon"])

plt.title("Survival by Gender")

plt.show()
```

##  Survival by Gender Insight

First-class passengers had a higher survival rate than second- and third-class passengers.

```python
fare = df["Fare"].sort_values()

plt.plot(fare, color="purple")

plt.fill_between(
    range(len(fare)),
    fare,
    color="plum",
    alpha=0.3
)

plt.title("Fare Distribution")

plt.xlabel("Passengers")

plt.ylabel("Fare")

plt.show()
```

# 💰 Fare Distribution Insight

Most passengers paid low fares, while a small number of passengers paid extremely high fares.

# Key Insights and Conclusions

## Main Findings

- Female passengers had a much higher survival rate than male passengers.

- Passengers in higher classes generally had better survival chances.

- Most passengers were young or middle-aged adults, while children and elderly passengers were less common.

- Most passengers paid relatively low fares, while a small number paid very high fares.

- Age and Fare did not show a strong direct relationship.

- Passengers traveling with a small family generally had better survival chances than passengers traveling alone or with very large families.

- Gender, passenger class, fare, and family size were important factors related to survival.

## Conclusion

The Titanic dataset shows that survival was influenced by several passenger characteristics. Gender and passenger class showed the clearest differences in survival rates, while fare, age, and family size also revealed useful patterns.

Through data cleaning, exploratory data analysis, and visualization, we were able to identify important patterns and relationships in the dataset and gain a better understanding of the factors related to passenger survival.
